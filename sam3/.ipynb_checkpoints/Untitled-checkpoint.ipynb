```python
import cv2
import numpy as np


def overlay_masks(
    frame, masks, alpha=0.4, draw_borders=True, show_ids=True, font_scale=0.7
):
    """Legt SAM 2 Masken halbtransparent über den Frame und zeichnet die Objekt-IDs in die Mitte."""
    output = frame.copy()
    h_frame, w_frame = frame.shape[:2]
    mask_dict = {}

    # 1. Masken entpacken
    if isinstance(masks, dict) and "out_obj_ids" in masks:
        obj_ids = masks["out_obj_ids"]
        mask_data = masks.get(
            "out_binary_masks",
            masks.get("out_mask_logits", masks.get("out_masks", None)),
        )

        if mask_data is not None:
            if hasattr(mask_data, "cpu"):
                mask_data = mask_data.cpu().numpy()
            if hasattr(obj_ids, "cpu"):
                obj_ids = obj_ids.cpu().numpy()

            if np.issubdtype(mask_data.dtype, np.floating):
                mask_data = mask_data > 0.0
            else:
                mask_data = mask_data.astype(bool)

            mask_data = np.squeeze(mask_data)
            if mask_data.ndim == 2:
                mask_data = np.expand_dims(mask_data, axis=0)

            for i, obj_id in enumerate(obj_ids):
                m = mask_data[i]
                if m.shape[:2] != (h_frame, w_frame):
                    m = cv2.resize(
                        m.astype(np.uint8),
                        (w_frame, h_frame),
                        interpolation=cv2.INTER_NEAREST,
                    ).astype(bool)
                mask_dict[obj_id] = m
    else:
        mask_dict = masks

    # 2. Farb-Generator
    def get_color(obj_id):
        seed_val = abs(hash(str(obj_id))) % 100000
        np.random.seed(seed_val)
        return np.random.randint(50, 255, size=3, dtype=np.uint8)

    # 3. Masken & IDs zeichnen
    for obj_id, mask in mask_dict.items():
        if mask is None:
            continue

        binary_mask = mask.astype(bool)
        if not np.any(binary_mask):
            continue

        color = get_color(obj_id)
        mask_uint8 = binary_mask.astype(np.uint8)

        # A) Halbtransparente Fläche
        output[binary_mask] = (
            (1 - alpha) * output[binary_mask] + alpha * color
        ).astype(np.uint8)

        # B) Konturlinie
        if draw_borders:
            contours, _ = cv2.findContours(
                mask_uint8, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
            )
            cv2.drawContours(output, contours, -1, color.tolist(), thickness=2)

        # C) ID in die Mitte der Maske schreiben
        if show_ids:
            # Schwerpunkt (Mitte) der Maske mittels Momenten berechnen
            M = cv2.moments(mask_uint8)
            if M["m00"] != 0:
                cX = int(M["m10"] / M["m00"])
                cY = int(M["m01"] / M["m00"])
            else:
                # Fallback: Mittelpunkt der Bounding Box
                ys, xs = np.where(binary_mask)
                cX, cY = int(xs.mean()), int(ys.mean())

            text = f"ID: {obj_id}"
            font = cv2.FONT_HERSHEY_SIMPLEX

            # Textgröße berechnen, um ihn perfekt zu zentrieren
            (text_w, text_h), _ = cv2.getTextSize(
                text, font, font_scale, thickness=2
            )
            text_x = cX - (text_w // 2)
            text_y = cY + (text_h // 2)

            # Schwarze Kontur für bessere Lesbarkeit auf jedem Hintergrund
            cv2.putText(
                output,
                text,
                (text_x, text_y),
                font,
                font_scale,
                (0, 0, 0),
                thickness=4,
                lineType=cv2.LINE_AA,
            )
            # Weißer Haupttext
            cv2.putText(
                output,
                text,
                (text_x, text_y),
                font,
                font_scale,
                (255, 255, 255),
                thickness=2,
                lineType=cv2.LINE_AA,
            )

    return output
```

```python
import cv2
from IPython.display import display
from ipyevents import Event
import ipywidgets as widgets

# Globaler Speicher für den letzten Klick
last_click = (
    None  # Speichert ein Dict: {'frame_idx': int, 'x': int, 'y': int}
)

H_orig, W_orig = video_frames[0].shape[:2]
DISPLAY_WIDTH = 800
DISPLAY_HEIGHT = int(DISPLAY_WIDTH * (H_orig / W_orig))

if isinstance(outputs_per_frame, dict):
    frame_keys = list(outputs_per_frame.keys())
else:
    frame_keys = list(range(len(outputs_per_frame)))

total_frames = len(frame_keys)

# ---------------------------------------------------------
# 1. UI-Widgets erstellen
# ---------------------------------------------------------
btn_prev = widgets.Button(
    description="◀ Zurück",
    button_style="primary",
    layout=widgets.Layout(width="120px"),
)
btn_next = widgets.Button(
    description="Weiter ▶",
    button_style="primary",
    layout=widgets.Layout(width="120px"),
)
slider = widgets.IntSlider(
    value=0,
    min=0,
    max=total_frames - 1,
    step=1,
    description="Frame:",
    continuous_update=True,
    layout=widgets.Layout(width="400px"),
)

coord_label = widgets.HTML(
    value="<b style='color:gray;'>Geklickte Koordinate:</b> <i>Klicke auf eine Stelle im Bild...</i>"
)

image_widget = widgets.Image(
    format="jpeg", width=DISPLAY_WIDTH, height=DISPLAY_HEIGHT
)


# ---------------------------------------------------------
# 2. Rendering-Funktion (rendert Frame + Masken + "X"-Marker)
# ---------------------------------------------------------
def update_display():
    idx = slider.value
    key = frame_keys[idx]

    # Frame mit Masken und IDs rendern
    rendered_rgb = overlay_masks(
        video_frames[key],
        outputs_per_frame[key],
        alpha=0.4,
        draw_borders=True,
        show_ids=True,
    )

    # Falls für DIESEN Frame ein Klick existiert -> "X" einzeichnen
    if last_click is not None and last_click["frame_idx"] == idx:
        cx, cy = last_click["x"], last_click["y"]

        # 1. Schwarze fette Außenkontur für das X (damit man es überall sieht)
        cv2.drawMarker(
            rendered_rgb,
            (cx, cy),
            (0, 0, 0),
            markerType=cv2.MARKER_TILTED_CROSS,
            markerSize=22,
            thickness=5,
            line_type=cv2.LINE_AA,
        )

        # 2. Knallgelbes X im Vordergrund
        cv2.drawMarker(
            rendered_rgb,
            (cx, cy),
            (255, 255, 0),
            markerType=cv2.MARKER_TILTED_CROSS,
            markerSize=20,
            thickness=2,
            line_type=cv2.LINE_AA,
        )

    # In JPEG-Bytes umwandeln und im Widget anzeigen
    rendered_bgr = cv2.cvtColor(rendered_rgb, cv2.COLOR_RGB2BGR)
    _, encoded_jpg = cv2.imencode(
        ".jpg", rendered_bgr, [int(cv2.IMWRITE_JPEG_QUALITY), 90]
    )
    image_widget.value = encoded_jpg.tobytes()


# ---------------------------------------------------------
# 3. Klick-Handler
# ---------------------------------------------------------
def handle_click(event):
    global last_click

    click_x = event.get("relativeX", 0)
    click_y = event.get("relativeY", 0)

    # Umrechnung auf Original-Auflösung
    real_x = int(click_x * (W_orig / DISPLAY_WIDTH))
    real_y = int(click_y * (H_orig / DISPLAY_HEIGHT))

    real_x = max(0, min(W_orig - 1, real_x))
    real_y = max(0, min(H_orig - 1, real_y))

    # Klick für den aktuellen Frame speichern
    last_click = {"frame_idx": slider.value, "x": real_x, "y": real_y}

    # Text-Anzeige aktualisieren
    coord_label.value = (
        f"<b style='font-size:15px; color:#1b5e20;'>Frame {slider.value} | "
        f"Geklickte Koordinate: <span style='color:#d9534f; font-size:16px;'>X = {real_x}, Y = {real_y}</span></b>"
    )

    # Anzeigebild mit dem neuen X sofort aktualisieren
    update_display()


# Event verknüpfen
im_event = Event(
    source=image_widget, watched_events=["click"], prevent_default_action=True
)
im_event.on_dom_event(handle_click)


# ---------------------------------------------------------
# 4. Steuerung & Initialisierung
# ---------------------------------------------------------
def on_prev_clicked(b):
    if slider.value > 0:
        slider.value -= 1


def on_next_clicked(b):
    if slider.value < total_frames - 1:
        slider.value += 1


def on_slider_change(change):
    update_display()


btn_prev.on_click(on_prev_clicked)
btn_next.on_click(on_next_clicked)
slider.observe(on_slider_change, names="value")

# UI Layout anzeigen
controls = widgets.HBox([btn_prev, slider, btn_next])
display(widgets.VBox([controls, coord_label, image_widget]))

# Ersten Frame zeichnen
update_display()
```

```python
def register_masks_to_predictor(
    predictor,
    inference_state,
    outputs_per_frame,
    ids_to_correct,
    verbose=True,
):
    for frame_idx in sorted(outputs_per_frame.keys()):
        frame_data = outputs_per_frame[frame_idx]
        raw_ids = frame_data["out_obj_ids"]
        raw_masks = frame_data["out_binary_masks"]

        # Tensoren zu NumPy umwandeln
        if hasattr(raw_ids, "cpu"):
            raw_ids = raw_ids.cpu().numpy()
        if hasattr(raw_masks, "cpu"):
            raw_masks = raw_masks.cpu().numpy()

        obj_ids_list = [int(x) for x in raw_ids]

        # -----------------------------------------------------------------
        # DIESE ZEILEN GEHÖREN HIER HIN (masks_arr aus raw_masks erstellen):
        # -----------------------------------------------------------------
        masks_arr = np.squeeze(raw_masks)
        if masks_arr.ndim == 2:
            masks_arr = np.expand_dims(masks_arr, axis=0)

        for obj_id in ids_to_correct:
            if obj_id in obj_ids_list:
                mask_index = obj_ids_list.index(obj_id)

                # Jetzt greift er auf die echte Maske DES AKTUELLEN FRAMES zu!
                single_mask = masks_arr[mask_index]

                # In PyTorch Tensor umwandeln
                single_mask_tensor = torch.from_numpy(single_mask)

                # An SAM 3 übergeben
                predictor.add_new_mask(
                    inference_state=inference_state,
                    frame_idx=frame_idx,
                    obj_id=obj_id,
                    mask=single_mask_tensor,
                    add_mask_to_memory=False,
                )

                if verbose:
                    print(
                        f"✔ Frame {frame_idx}: Maske für ID {obj_id} übergeben."
                    )
```

```python
import cv2
from IPython.display import display
from ipyevents import Event
import ipywidgets as widgets
import numpy as np
import torch

# Abmessungen des Videos
H_orig, W_orig = video_frames[0].shape[:2]
width, height = W_orig, H_orig
DISPLAY_WIDTH = 800
DISPLAY_HEIGHT = int(DISPLAY_WIDTH * (H_orig / W_orig))

# NEU: Dictionary speichert Klicks gruppiert nach Objekt-ID
# Struktur: { obj_id: {'points': [[x,y], ...], 'labels': [1, 0, ...]} }
collected_clicks = {}

if isinstance(video_segments, dict):
    frame_keys = sorted(list(video_segments.keys()))
else:
    frame_keys = list(range(len(video_segments)))

total_frames = len(frame_keys)

# ---------------------------------------------------------
# 1. Widgets erstellen
# ---------------------------------------------------------
btn_prev = widgets.Button(
    description="◀ Zurück",
    button_style="primary",
    layout=widgets.Layout(width="100px"),
)
btn_next = widgets.Button(
    description="Weiter ▶",
    button_style="primary",
    layout=widgets.Layout(width="100px"),
)
slider = widgets.IntSlider(
    value=0,
    min=0,
    max=total_frames - 1,
    step=1,
    description="Frame:",
    continuous_update=True,
    layout=widgets.Layout(width="350px"),
)

obj_id_dropdown = widgets.Dropdown(
    options=[], description="Ziel-ID:", layout=widgets.Layout(width="160px")
)

click_type_toggle = widgets.ToggleButtons(
    options=[("➕ Positiv", 1), ("➖ Negativ", 0)],
    description="Klick-Typ:",
    button_style="",
    layout=widgets.Layout(width="220px"),
)

btn_reset_clicks = widgets.Button(
    description="🗑️ Klicks löschen",
    button_style="warning",
    layout=widgets.Layout(width="140px"),
)

btn_submit_refinement = widgets.Button(
    description="🚀 Submit & Re-Track",
    button_style="success",
    layout=widgets.Layout(width="180px"),
)

status_label = widgets.HTML(
    value="<b>Status:</b> 0 Klicks gesammelt. <i>Wähle eine ID und klicke ins Bild...</i>"
)

image_widget = widgets.Image(
    format="jpeg", width=DISPLAY_WIDTH, height=DISPLAY_HEIGHT
)


# ---------------------------------------------------------
# 2. Hilfsfunktionen
# ---------------------------------------------------------
def update_id_dropdown():
    idx = slider.value
    key = frame_keys[idx]

    current_ids = []
    if key in video_segments and isinstance(video_segments[key], dict):
        current_ids = sorted(list(video_segments[key].keys()))

    if current_ids:
        obj_id_dropdown.options = [str(i) for i in current_ids]
    else:
        obj_id_dropdown.options = ["19"]


def update_display():
    """Zeichnet Frame + Masken + ALLE Klick-Punkte (inkl. ID-Text)."""
    idx = slider.value
    key = frame_keys[idx]

    rendered_rgb = overlay_masks(
        video_frames[key],
        video_segments[key],
        alpha=0.4,
        draw_borders=True,
        show_ids=True,
    )

    # Alle gesammelten Klicks aller Objekt-IDs zeichnen
    for obj_id, click_data in collected_clicks.items():
        for pt, lbl in zip(click_data["points"], click_data["labels"]):
            cx, cy = pt[0], pt[1]
            marker_color = (0, 255, 0) if lbl == 1 else (0, 0, 255)

            # Punkt zeichnen
            cv2.circle(
                rendered_rgb, (cx, cy), 7, (0, 0, 0), -1, lineType=cv2.LINE_AA
            )
            cv2.circle(
                rendered_rgb,
                (cx, cy),
                5,
                marker_color,
                -1,
                lineType=cv2.LINE_AA,
            )

            # Zugehörige Objekt-ID direkt neben dem Punkt anzeigen!
            cv2.putText(
                rendered_rgb,
                str(obj_id),
                (cx + 8, cy + 4),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 0, 0),
                3,
                lineType=cv2.LINE_AA,
            )
            cv2.putText(
                rendered_rgb,
                str(obj_id),
                (cx + 8, cy + 4),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 255, 255),
                1,
                lineType=cv2.LINE_AA,
            )

    rendered_bgr = cv2.cvtColor(rendered_rgb, cv2.COLOR_RGB2BGR)
    _, encoded_jpg = cv2.imencode(
        ".jpg", rendered_bgr, [int(cv2.IMWRITE_JPEG_QUALITY), 90]
    )
    image_widget.value = encoded_jpg.tobytes()


# ---------------------------------------------------------
# 3. Event-Handlers
# ---------------------------------------------------------
def handle_click(event):
    """Speichert Klicks genau zugeordnet zur aktuell gewählten Objekt-ID."""
    click_x = event.get("relativeX", 0)
    click_y = event.get("relativeY", 0)

    real_x = int(click_x * (W_orig / DISPLAY_WIDTH))
    real_y = int(click_y * (H_orig / DISPLAY_HEIGHT))

    real_x = max(0, min(W_orig - 1, real_x))
    real_y = max(0, min(H_orig - 1, real_y))

    # Ziel-ID aus Dropdown holen
    try:
        target_obj_id = int(obj_id_dropdown.value)
    except (ValueError, TypeError):
        target_obj_id = 19

    # Falls für diese ID noch keine Klicks existieren -> anlegen
    if target_obj_id not in collected_clicks:
        collected_clicks[target_obj_id] = {"points": [], "labels": []}

    current_label = click_type_toggle.value
    collected_clicks[target_obj_id]["points"].append([real_x, real_y])
    collected_clicks[target_obj_id]["labels"].append(current_label)

    # Zusammenfassung für den Status-Text erstellen
    summary = ", ".join(
        f"ID {oid}: {len(data['points'])} Klicks"
        for oid, data in collected_clicks.items()
    )
    status_label.value = f"<b>Gesammelte Klicks:</b> {summary}"

    update_display()


def on_reset_clicked(b):
    """Löscht alle Klicks aller Objekt-IDs."""
    collected_clicks.clear()
    status_label.value = (
        "<b>Status:</b> Klicks gelöscht. <i>Klicke neu ins Bild...</i>"
    )
    update_display()


def on_submit_clicked(b):
    """Übergibt die Klicks für ALLE Objekt-IDs an SAM 3 und propagiert im Video."""
    global video_segments

    if not collected_clicks:
        status_label.value = (
            "<b style='color:red;'>⚠️ Bitte erst Klicks im Bild setzen!</b>"
        )
        return

    frame_to_correct = slider.value
    ids_list_str = ", ".join(str(i) for i in collected_clicks.keys())
    status_label.value = f"<b style='color:blue;'>🔄 SAM 3 Refinement für IDs [{ids_list_str}] läuft... Bitte warten!</b>"

    try:
        device_type = "cuda" if torch.cuda.is_available() else "cpu"
        target_dtype = (
            torch.bfloat16 if torch.cuda.is_available() else torch.float32
        )

        with torch.inference_mode():
            with torch.autocast(device_type=device_type, dtype=target_dtype):
                # 1. Schleife über ALLE gesammelten Objekt-IDs
                for obj_id, click_data in collected_clicks.items():
                    rel_points = [
                        [x / width, y / height] for x, y in click_data["points"]
                    ]

                    points_tensor = torch.tensor(
                        rel_points, dtype=target_dtype
                    ).unsqueeze(0)
                    labels_tensor = torch.tensor(
                        click_data["labels"], dtype=torch.int32
                    ).unsqueeze(0)

                    # Für jede ID separat add_new_points_or_box aufrufen
                    predictor.add_new_points_or_box(
                        inference_state=inference_state,
                        frame_idx=frame_to_correct,
                        obj_id=obj_id,
                        points=points_tensor,
                        labels=labels_tensor,
                        clear_old_points=True,
                        normalize_coords=True,
                    )

                # 2. Propagierung EINMALIG fürs ganze Video durchführen
                new_video_segments = {}
                for (
                    f_idx,
                    obj_ids,
                    l_masks,
                    v_masks,
                    scores,
                ) in predictor.propagate_in_video(
                    inference_state,
                    start_frame_idx=0,
                    max_frame_num_to_track=total_frames,
                    reverse=False,
                    propagate_preflight=True,
                ):
                    new_video_segments[f_idx] = {
                        out_obj_id: np.squeeze(
                            (v_masks[i] > 0.0).clone().cpu().numpy()
                        )
                        for i, out_obj_id in enumerate(obj_ids)
                    }

        # 3. Aktualisieren & Klicks zurücksetzen
        video_segments = new_video_segments
        collected_clicks.clear()

        status_label.value = (
            f"<b style='color:green;'>✅ Erfolgreich! IDs [{ids_list_str}] ab Frame {frame_to_correct} "
            f"refined und propagiert.</b>"
        )

    except Exception as e:
        status_label.value = (
            f"<b style='color:red;'>❌ Fehler bei SAM 3 Refinement: {str(e)}</b>"
        )

    update_display()


# Events verknüpfen
im_event = Event(
    source=image_widget, watched_events=["click"], prevent_default_action=True
)
im_event.on_dom_event(handle_click)

btn_reset_clicks.on_click(on_reset_clicked)
btn_submit_refinement.on_click(on_submit_clicked)


def on_prev_clicked(b):
    if slider.value > 0:
        slider.value -= 1


def on_next_clicked(b):
    if slider.value < total_frames - 1:
        slider.value += 1


def on_slider_change(change):
    update_id_dropdown()
    update_display()


btn_prev.on_click(on_prev_clicked)
btn_next.on_click(on_next_clicked)
slider.observe(on_slider_change, names="value")

# ---------------------------------------------------------
# 4. Layout aufbauen & Starten
# ---------------------------------------------------------
update_id_dropdown()

controls_nav = widgets.HBox([btn_prev, slider, btn_next])
controls_refine = widgets.HBox(
    [obj_id_dropdown, click_type_toggle, btn_reset_clicks, btn_submit_refinement]
)

display(widgets.VBox([controls_nav, controls_refine, status_label, image_widget]))

update_display()
```
