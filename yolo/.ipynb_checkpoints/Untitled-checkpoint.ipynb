```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# ================= PFADE ANPASSEN =================
# Wähle hier ein Bild und die passende txt-Datei aus deinem Dataset
IMG_PATH = Path("./dataset/images/train/video_001_0001_to_0040/0001.jpg")
LBL_PATH = Path("./dataset/labels/train/video_001_0001_to_0040/0001.txt")
# ==================================================

# 1. Bild laden
img = cv2.imread(str(IMG_PATH))
if img is None:
    raise FileNotFoundError(f"Bild nicht gefunden unter: {IMG_PATH}")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
h, w = img_rgb.shape[:2]

# Leere Maske (schwarz) und Overlay-Bild für das Plotten erstellen
pure_mask = np.zeros((h, w), dtype=np.uint8)
overlay_img = img_rgb.copy()

# 2. Labels aus TXT einlesen und Maske zeichnen
if LBL_PATH.exists():
    with open(LBL_PATH, "r") as f:
        lines = f.readlines()

    print(f"Gefundene Objekte im Frame: {len(lines)}")

    for line in lines:
        parts = line.strip().split()
        if not parts:
            continue

        class_id = parts[0]
        coords = [float(x) for x in parts[1:]]

        # Denormalisieren: (x_norm, y_norm) -> (x_pixel, y_pixel)
        pts = []
        for i in range(0, len(coords), 2):
            px = int(coords[i] * w)
            py = int(coords[i + 1] * h)
            pts.append([px, py])

        pts = np.array(pts, dtype=np.int32).reshape((-1, 1, 2))

        # A) Auf die reine Binärmaske zeichnen
        cv2.fillPoly(pure_mask, [pts], 255)

        # B) Auf das Overlay-Bild zeichnen (grün transparent + weiße Kontur)
        cv2.polylines(overlay_img, [pts], isClosed=True, color=(255, 255, 255), thickness=2)

    # Farbige Masken-Überlagerung für das Originalbild
    colored_mask = np.zeros_like(img_rgb)
    colored_mask[pure_mask == 255] = [0, 255, 0]  # Grün
    overlay_img = cv2.addWeighted(overlay_img, 0.7, colored_mask, 0.3, 0)

else:
    print(f"Keine Label-Datei gefunden: {LBL_PATH}")

# 3. Visualisierung via Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(14, 7))

# Linker Plot: Reine Maske
axes[0].imshow(pure_mask, cmap="gray")
axes[0].set_title(f"Extrahierte Maske\n({LBL_PATH.name})")
axes[0].axis("off")

# Rechter Plot: Overlay auf Originalbild
axes[1].imshow(overlay_img)
axes[1].set_title(f"Overlay auf Originalbild\n({IMG_PATH.name})")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
import torch
from ultralytics import YOLO

# 2. Prüfen, ob GPU (CUDA) verfügbar ist
device = 0 if torch.cuda.is_available() else "cpu"
print(f"Verwendetes Device: {'GPU (' + torch.cuda.get_device_name(0) + ')' if device == 0 else 'CPU'}")

# 3. Vortrainiertes Segmentierungs-Modell laden
# Optionen für Größe: yolov8n-seg (Nano), yolov8s-seg (Small), yolov8m-seg (Medium)
# Alternativ das neuere: yolo11n-seg.pt
model = YOLO("yolo26n-seg.pt")

# 4. Training starten
results = model.train(
    data="./dataset/data.yaml",
    epochs=50,
    imgsz=1024,
    rect=True,       # Hält das echte Hochkant-Format bei
    mosaic=0.0,      # <--- Deaktiviert das 4er-Mosaik komplett!
    mixup=0.0,       # Deaktiviert das Überblenden von Bildern
    degrees=5.0,     # Nur leichte Drehung
    batch=8
)

print("\n--- Training abgeschlossen! ---")
```
